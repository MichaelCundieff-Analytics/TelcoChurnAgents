# TelcoChurnAgents: Multi-Agent AI System for Telecom Customer Churn Prediction

A collaborative AI system where two specialized agents work together to predict customer churn and recommend retention strategies.

## Problem Statement

Telecom companies face significant customer churn, leading to revenue loss and increased acquisition costs. This project addresses this challenge by:

1. **Predicting churn risk** - Identifying which customers are most likely to leave
2. **Recommending retention strategies** - Suggesting targeted interventions for high-risk segments

The solution leverages a multi-agent architecture where different AI agents specialize in different aspects of the problem, demonstrating how collaboration can solve complex business challenges.

## The Two Agents

### Agent 1: Data Science Agent
**Role:** Technical machine learning pipeline execution

Responsibilities:
- Data loading and profiling (csv_importer.py)
- Data cleaning and preprocessing (data_cleaner.py)
- Exploratory data analysis (eda_analyzer.py)
- Model training and comparison (model_trainer.py)
- Model evaluation (model_evaluator.py)
- Customer risk scoring (risk_scorer.py)

**Output:** Processed datasets, trained models, customer risk scores, and feature importance metrics

### Agent 2: Retention Strategy Agent
**Role:** Business intelligence and retention strategy formulation

Responsibilities:
- Analyzing high-risk customer segments (segment_analyzer.py)
- Generating retention recommendations (retention_recommender.py)
- Creating final report and dashboard (output_generator.py)

**Output:** Actionable business recommendations, segment insights, and executive dashboard

### How They Work Together
1. Data Science Agent completes full ML pipeline and writes handoff summary
2. Retention Strategy Agent reads all outputs and generates business recommendations
3. Final collaboration produces both technical metrics and business insights

## Setup

```python
import os
import json
import pandas as pd
import sys

# Set project root
project_root = os.path.dirname(os.path.dirname(os.path.abspath('__file__')))
print(f'Project root: {project_root}')
print(f'Current directory: {os.getcwd()}')
```

## Run the Multi-Agent Orchestrator

Execute the complete pipeline with both agents:

```python
import subprocess
import sys
from pathlib import Path

# Find project root by looking for skills/orchestrator.py
start_dir = Path.cwd().resolve()
project_root = None

for candidate in [start_dir] + list(start_dir.parents):
    if (candidate / "skills" / "orchestrator.py").exists():
        project_root = candidate
        break

if project_root is None:
    raise FileNotFoundError("Could not find skills/orchestrator.py. Make sure this notebook is inside your TelcoChurnAgents project folder.")

print(f"Project root: {project_root}")

# Run the orchestrator which coordinates both agents
result = subprocess.run(
    [sys.executable, "skills/orchestrator.py"],
    capture_output=True,
    text=True,
    cwd=str(project_root)
)

print("STDOUT:")
print(result.stdout)

print("\nSTDERR:")
print(result.stderr)

print(f"\nReturn code: {result.returncode}")

if result.returncode != 0:
    raise RuntimeError("The orchestrator failed. Check STDERR above.")
```

## Customer Risk Scores

```python
import pandas as pd

risk_scores_path = project_root / "data" / "processed" / "customer_risk_scores.csv"

if risk_scores_path.exists():
    df_risk = pd.read_csv(risk_scores_path)

    print(f"Total customers scored: {len(df_risk)}")

    print("\nRisk level distribution:")
    print(df_risk["risk_level"].value_counts())

    print("\nChurn probability summary:")
    print(df_risk["churn_probability"].describe())

    print("\nFirst 10 customers:")
    display(df_risk.head(10))
else:
    print(f"Risk scores file not found at {risk_scores_path}")
```

## How the Agents Worked Together

### The Collaboration Process

**Phase 1: Data Science Agent (Technical Foundation)**
- Imported raw customer data from CSV
- Cleaned and preprocessed data for ML readiness
- Performed exploratory data analysis to understand patterns
- Trained multiple ML models to predict churn likelihood
- Evaluated models and selected best performer
- Scored all customers with risk predictions
- Calculated feature importance to identify key drivers
- Wrote comprehensive handoff summary with all findings

**Phase 2: Retention Strategy Agent (Business Insights)**
- Read all technical outputs from Data Science Agent
- Analyzed high-risk customer segments using the risk scores
- Identified patterns and characteristics of churning customers
- Generated targeted retention recommendations
- Prioritized interventions by potential impact
- Created executive dashboard and final report
- Translated technical metrics into actionable business strategies

### Why Multi-Agent Architecture?

1. **Separation of Concerns** - Each agent focuses on its domain expertise
2. **Scalability** - Agents can be independently updated or replaced
3. **Clarity** - Clear handoff points make the system transparent and auditable
4. **Specialization** - Technical and business intelligence handled by specialized agents
5. **Collaboration** - File-based communication ensures reproducibility

The result is a system that not only predicts churn with high accuracy but also provides business teams with clear, actionable recommendations for customer retention—demonstrating how AI agents with different roles solve problems neither could handle alone.
